# ColumnLayer Example

This is a quick example notebook to show how to use the `ColumnLayer`.

This is derived from [`@cboettig`](https://github.com/cboettig)'s example in [this issue](https://github.com/developmentseed/lonboard/issues/543).

![image.png](../assets/column-layer.jpg)


(The `uvx` command is included when installing `uv`).


## Imports


````python
import geopandas as gpd

from lonboard import ColumnLayer, Map
````

````python
DATA_URL = "https://data.source.coop/cboettig/conservation-policy/Inflation_Reduction_Act_Projects.geojson"
````

````python
df = gpd.read_file(DATA_URL, use_arrow=True)
````

````python
layer = ColumnLayer.from_geopandas(
    df,
    get_elevation=df["FUNDING_NUMERIC"],
    get_fill_color=[255, 255, 0, 140],
    elevation_scale=0.01,
    radius=10000,
    pickable=True,
    auto_highlight=True,
)
````

````python
m = Map(
    layer,
    view_state={
        "longitude": -100,
        "latitude": 35,
        "zoom": 4,
        "pitch": 45,
        "bearing": 0,
    },
)
m
````
